```python
def perceptron(X,y):
    
    m = []
    b = []
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    
    for i in range(200):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
        
        m.append(-(weights[1]/weights[2]))
        b.append(-(weights[0]/weights[2]))
        
    return m,b
```

```python
def step(z):
    return 1 if z>0 else 0 
    
```

```python
def perpetron(X,y):
    X=np.insert(X,0,1,axis=1) # for first column having all 1 
    weights=np.ones(X.shape[1]) # now new no. of columns is 3 and initially we are putting w0,w1,w2 = 1 
    lr = 0.1
    for i in range(200):
        j = np.random.randint(0,100)# cause our X has 100 rows or its shape is (100,2) now 
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
    return weights[0],weights[1:]
        
```

```python
intercept_,coef_=perpetron(X,y)
```

```python
print(coef_) # w1 ,w2 
print(intercept_)# w0
```
