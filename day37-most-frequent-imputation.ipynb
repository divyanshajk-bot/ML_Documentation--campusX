```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df=pd.read_csv('train1.csv',usecols=['GarageQual','FireplaceQu','SalePrice'])
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
df['GarageQual'].value_counts().plot(kind='bar')
```

```python
df['GarageQual'].mode()
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

df[df['GarageQual']=='TA']['SalePrice'].plot(kind='kde', ax=ax)

df[df['GarageQual'].isnull()]['SalePrice'].plot(kind='kde', ax=ax, color='red')

lines, labels = ax.get_legend_handles_labels()
labels = ['Houses with TA', 'Houses with NA']
ax.legend(lines, labels, loc='best')

plt.title('GarageQual')
```

```python
#saleprice ka pdf plot kar rhe hai jaha pe garage quality ka value TA  hai 
#saleprice ka pdf plot kar rhe hai jaha pe garage quality ka value null hai 
# dono ko side by side plot kar rhe hai 
```

```python
temp=df[df['GarageQual']=='TA']['SalePrice']
```

```python
df['GarageQual'].fillna('TA',inplace=True)
```

```python
# thus we will do 
df['GarageQual']=df['GarageQual'].fillna('TA')
```

```python
df['GarageQual'].value_counts().plot(kind='bar')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)


temp.plot(kind='kde', ax=ax)

# distribution of the variable after imputation
df[df['GarageQual'] == 'TA']['SalePrice'].plot(kind='kde', ax=ax, color='red')

lines, labels = ax.get_legend_handles_labels()
labels = ['Original variable', 'Imputed variable']
ax.legend(lines, labels, loc='best')

# add title
plt.title('GarageQual')
```

```python
df['FireplaceQu'].value_counts().plot(kind='bar')
```

```python
# here most frequent imputation will not work that good because any one category is not dominating much 
# as  u can see Gd and TA are almost equal and dont have much difference 
```

```python
df['FireplaceQu'].mode()
```

```python
fig=plt.figure()
ax=fig.add_subplot(111)

df[df['FireplaceQu']=='Gd']['SalePrice'].plot(kind='kde', ax=ax)

df[df['FireplaceQu'].isnull()]['SalePrice'].plot(kind='kde', ax=ax, color='red')

lines, labels = ax.get_legend_handles_labels()
labels = ['Houses with Gd', 'Houses with NA']
ax.legend(lines, labels, loc='best')

plt.title('FireplaceQu')
```

```python
temp = df[df['FireplaceQu']=='Gd']['SalePrice']
```

```python
df['FireplaceQu'].fillna('Gd', inplace=True)
```

```python
df['FireplaceQu'].value_counts().plot(kind='bar')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)


temp.plot(kind='kde', ax=ax)

# distribution of the variable after imputation
df[df['FireplaceQu'] == 'Gd']['SalePrice'].plot(kind='kde', ax=ax, color='red')

lines, labels = ax.get_legend_handles_labels()
labels = ['Original variable', 'Imputed variable']
ax.legend(lines, labels, loc='best')

# add title
plt.title('FireplaceQu')
```

```python
# lots of missing values 
# koi ek category dominate nhi kar rha tha 
```

```python
# conclusion : we should not use most frequent imputation in this column as it wont give u good results 
```
