```python
#sns.distplot(df['Age'])
sns.histplot(df['Age'],kde=True)
#  the line curve drawn in the plot is KDE : kernel density estimation  
# this curve tells us the distribution of data and also called PDF : probability density function 
# probability of any person from this group of being of a certain age will be given by corresponding point of given age ka y coordinate 
```
