```python
import numpy as np # linear algebra
import pandas as pd # data processing
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('wine_data.csv',header=None,usecols=[0,1,2])
df.columns=['Class label', 'Alcohol', 'Malic acid']
```

```python
df
```

```python
sns.kdeplot(df['Alcohol'])
```

```python
sns.kdeplot(df['Malic acid'])
```

```python
color_dict={1:'red',2:'blue',3:'green'}
```

```python
sns.scatterplot(x=df['Alcohol'],y=df['Malic acid'],hue=df['Class label'],palette=color_dict)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop('Class label', axis=1),
                                                    df['Class label'],
                                                    test_size=0.3,
                                                    random_state=0)

X_train.shape, X_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
## Key Characteristics

#1. **Parameter Learning**:
   #- Scans through all samples in the dataset
   #- Finds min/max values for each feature column
   #- Stores these parameters internally for later use

#2. **No Data Transformation**:
  # - `fit()` does NOT change or return transformed data
  # - It only learns and stores parameters
  # - Returns the scaler object itself (for method chaining)

#3. **Feature-wise Analysis**:
  # - Analyzes each feature (column) independently
  # - Calculates statistics across all samples for each feature


# fit the scaler to the train set, it will learn the parameters
scaler.fit(X_train)

# transform train and test sets
#The `transform()` function applies the scaling transformation to data using the parameters that were previously learned during the `fit()` step.
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#to convert numpy array created due to sklearn to pandas dATAFRAME
X_train_scaled=pd.DataFrame(X_train_scaled,columns=X_train.columns)
X_test_scaled=pd.DataFrame(X_test_scaled,columns=X_test.columns)
```

```python
np.round(X_train.describe(), 1)
```

```python
np.round(X_train_scaled.describe(),1)
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

ax1.scatter(X_train['Alcohol'], X_train['Malic acid'],c=y_train)
ax1.set_title("Before Scaling")
ax2.scatter(X_train_scaled['Alcohol'], X_train_scaled['Malic acid'],c=y_train)
ax2.set_title("After Scaling")
plt.show()
```

```python
fig , (ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))

#before scaling
ax1.set_title('Before Scaling')
sns.kdeplot(X_train['Alcohol'], ax=ax1)
sns.kdeplot(X_train['Malic acid'], ax=ax1)

# after scaling
ax2.set_title('After Standard Scaling')
sns.kdeplot(X_train_scaled['Alcohol'], ax=ax2)
sns.kdeplot(X_train_scaled['Malic acid'], ax=ax2)
plt.show()
```

```python
fig , (ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))

#before scaling 
ax1.set_title('Malic Acid distribution before scaling ')
sns.kdeplot(X_train_scaled['Malic acid'],ax=ax1)

#after scaling 
ax2.set_title('Malic Acid distribution after scaling ')
sns.kdeplot(X_train_scaled['Malic acid'],ax=ax2)
plt.show()
```

```python
fig , (ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))

#before scaling 
ax1.set_title('Alcohol distribution before scaling ')
sns.kdeplot(X_train_scaled['Alcohol'],ax=ax1)

#after scaling 
ax2.set_title('Alcohol distribution after scaling ')
sns.kdeplot(X_train_scaled['Alcohol'],ax=ax2)
plt.show()
```
