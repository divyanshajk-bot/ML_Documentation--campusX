```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
```

```python
df=pd.read_csv('train.csv')
```

```python
df.sample(6)
```

```python
df.drop(columns=['PassengerId','Name','Ticket','Cabin'],inplace=True)
```

```python
df.sample(5)
```

```python
#Step 1 --> train/test/split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['Survived']),df['Survived'],test_size=0.2,random_state=42)
```

```python
# Applying imputation 
#si_age=SimpleImputer()
#si_embarked=SimpleImputer(strategy='most_frequent')# mode se replace kr diya missing values ko 

#X_train_age=si_age.fit_transform(X_train['Age']) # gives numpy array 
#X_train_embarked=si_embarked.fit_transform(X_train['Embarked'])

#X_test_age=si_age.fit_transform(X_test['Age'])
#X_test_embarked=si_embarked.fit_transform(X_test['Embarked'])

#This error occurs because `SimpleImputer` expects a 2-dimensional array (like a DataFrame or 2D numpy array), but you're passing a pandas Series (which is 1-dimensional). When you select a single column from a DataFrame using `X_train['Age']`, it returns a Series instead of a DataFrame.

#The solution is to pass the columns as DataFrames instead of Series by using double square brackets `[['column_name']]` instead of single brackets `['column_name']`.
```

```python
si_age=SimpleImputer()
si_embarked=SimpleImputer(strategy='most_frequent')# mode se replace kr diya missing values ko 

X_train_age=si_age.fit_transform(X_train[['Age']]) # gives numpy array 
X_train_embarked=si_embarked.fit_transform(X_train[['Embarked']])

X_test_age=si_age.fit_transform(X_test[['Age']])
X_test_embarked=si_embarked.fit_transform(X_test[['Embarked']])
```

```python
X_train_age
```

```python
X_train_embarked
```

```python
# one hot encoding in sex and embarked 

ohe_sex=OneHotEncoder(sparse_output=False,handle_unknown='ignore')
ohe_embarked = OneHotEncoder(sparse_output=False,handle_unknown='ignore')

X_train_sex=ohe_sex.fit_transform(X_train[['Sex']])
X_train_embarked=ohe_embarked.fit_transform(X_train_embarked)

X_test_sex=ohe_sex.transform(X_test[['Sex']])
X_test_embarked=ohe_embarked.transform(X_test_embarked)
```

```python
X_train_embarked
```

```python
X_train_rem=X_train.drop(columns=['Sex','Age','Embarked'])
```

```python
X_test_rem=X_test.drop(columns=['Sex','Age','Embarked'])
```

```python
X_train_transformed=np.concatenate((X_train_rem,X_train_age,X_train_sex,X_train_embarked),axis=1)
X_test_transformed=np.concatenate((X_test_rem,X_test_age,X_test_sex,X_test_embarked),axis=1)
```

```python
X_train_transformed
```

```python
X_test_transformed
```

```python
#model train
clf=DecisionTreeClassifier()
clf.fit(X_train_transformed,y_train)
```

```python
y_pred=clf.predict(X_test_transformed)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
